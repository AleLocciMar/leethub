```python
from typing import List

class SolutionDS:
    def generateParenthesis(self, n: int) -> List[str]:
        result = []
        
        def backtrack(combinacao: str, abertos: int, fechados: int):
            # Se temos n abertos e n fechados, completamos uma combinação
            if len(combinacao) == 2 * n:
                result.append(combinacao)
                return
            
            # Podemos adicionar '(' se ainda não usamos todos
            if abertos < n:
                backtrack(combinacao + '(', abertos + 1, fechados)
            
            # Podemos adicionar ')' se temos mais abertos que fechados
            if fechados < abertos:
                backtrack(combinacao + ')', abertos, fechados + 1)
        
        backtrack('', 0, 0)
        return result
```

```python
solDS= SolutionDS()
print(solDS.generateParenthesis(100))
```
