```python
class Solution:
    
    def myAtoi(self, s: str) -> int:
       s= s.strip()
       if s=="+" or s== "-":
           return 0
        
       p= 0
       e= len(s)        
       find= True
       
       while(p < e):
          if (s[p]=="+" or s[p]== "-") and find:
              p+=1
              find= False
          elif s[p].isdecimal():             
              p+=1
          else:
             break

       if p== 0:
           return 0
       else:
           out=s[:p] 

       if out=="-" or out=="+":
            return 0           
       if int(out) <= -2**31:
           return -2**31
       elif int(out) >= 2**31 -1 :
           return 2**31 - 1
           
       return int(out)  
        
```

```python
sol = Solution()
```

```python
print(sol.myAtoi("21474836460"))
```

```python
print(sol.myAtoi("-00045"))
print(sol.myAtoi("-1337+c0+d3"))
print(sol.myAtoi("-0-1"))
print(sol.myAtoi("words and 987"))
print(sol.myAtoi("-91283472332"))
print(sol.myAtoi("91283472332"))
print(sol.myAtoi("  .1"))
print(sol.myAtoi("+0-1"))
print(sol.myAtoi("---"))
print(sol.myAtoi("+-12"))
print(sol.myAtoi("9128347c2332"))
print(sol.myAtoi("3.14159"))
print(sol.myAtoi("+"))
```
